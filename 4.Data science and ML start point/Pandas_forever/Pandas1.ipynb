START PANDAS SERIES

```python
sales=pd.Series([5,6,2,3,5,0,9])
print(sales)
#to create and print a series use the above
```

```python
sales[1]
#to access a particular variable at a particular index we can call it like this in the series
```

```python
sales[0:2]
#data slicing refers to the slicing or retreiving certain part of the data ranging from 0th index position to <nth position as pre4sented in the above position 
```

```python
#ATTRIBUTES USED TO FIND INFORMATION ABOUT THE SPECIFIC SERIES
sales.index #.index is used here to provide the specific index values implemented/used by the series
sales.values #.values is used here to provide the specific values which the series conatins
sales.dtype #.dtype is used here to know what the specific datatype of the values of the series contains
```

METHODS USED TO SEE DATA IN SERIES

```python
#1.head(n)
sales.head(1)
#a .head() method is used to display the "HEAD" ie the starting (0th position) to the nth position(here n is the no. of elements you want to see from the start ie its the limit to where you want to see) 
```

```python
#2 .tail(n)
sales.tail(1)
#just think the opposite of tail i.e. starting from bottom ,rest properties kinda remain same(0th to n changes to the ending point to the upper levels)
```

```python
#3.sort_values()
sales.sort_values()
#if you think the data is not sorted use this
```

```python
#4.count()
sales.count()
#counts number of elements in this series  
```

```python
#5.isnull()
sales.isnull()
#used to see in data where the positions are blank i.e. null
```

```python
#6.notnull()
sales.notnull()
#used to see in data where the positions are not blank ; hence "notnull"
```
