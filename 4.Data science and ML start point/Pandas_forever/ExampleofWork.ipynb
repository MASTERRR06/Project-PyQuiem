```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv(r'/4.Data science and ML start point\Sample Dataset\ecommerce_product_dataset.csv')
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
print(df.dropna())
```

```python
df=df.fillna(0)
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df=df.drop_duplicates()
print(df.head())
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Sales'], bins=20)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
corr = df.corr(numeric_only=True)
print(corr['Sales'].sort_values(ascending=False))
```

```python
corr['Sales'].sort_values().plot(kind='barh')

plt.title("Correlation of Features with Sales")
plt.xlabel("Correlation")
plt.show()
```
